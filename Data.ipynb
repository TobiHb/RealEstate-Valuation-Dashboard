```python
import pandas as pd

# Import the dataset
file_path = "./NY_House_Dataset.csv"  # Adjust path if necessary
ny_house_dataset = pd.read_csv(file_path)

# Display the dataset
ny_house_dataset.head()
```

```python
columns_to_keep = ["PRICE", "BEDS", "BATH", "PROPERTYSQFT", "STATE", "TYPE"]

# Keep only selected columns in dataframe
df = ny_house_dataset[columns_to_keep]
df
```

```python
# Convert numerical values to int
columns_to_convert = ["PRICE", "BEDS", "BATH", "PROPERTYSQFT"]
df.loc[:, columns_to_convert] = df[columns_to_convert].apply(pd.to_numeric, errors='coerce').fillna(0).astype(int)
df
```

```python
# Use loc, for assigning the columns directly
# Extract only relevant strings
df.loc[:, 'STATE'] = df['STATE'].str.split().str[-1]  # Because ZIP is last string
df.loc[:, 'TYPE'] = df['TYPE'].str.split().str[0]  # Because Type is first string
df
```

```python
# For oversight
unique_type = df['TYPE'].unique()
print(unique_type)
```

```python
# For oversight
unique_state = df['STATE'].unique()
print(unique_state)
```

```python
# For oversight
# Sort unique values after numerical size for oversight
unique_state_sorted = sorted(unique_state, key=lambda x: int(x))

# Print sorted values
print(unique_state_sorted)
```

```python
# Execute One-hot encoding for the columns 'STATE' and 'TYPE'
df_encoded = pd.get_dummies(df, columns=['STATE', 'TYPE'], prefix=['STATE', 'TYPE'])
df_encoded
```

```python
print(df_encoded.shape)
```

```python
# Analysing upper outliers
top_10_prices = df_encoded["PRICE"].nlargest(20)
print(top_10_prices)
```

```python
# Removing upper outliers based on 99. percentile
def remove_upper_outliers(df, columns, percentile=0.99):
    for col in columns:
        upper_limit = df[col].quantile(percentile)
        df = df[df[col] <= upper_limit]
    return df


# Remove datapoint with upper outliers in df
df_encoded = remove_upper_outliers(df_encoded, ['PRICE'])

# Statistic of cleaned dataframe
print(df_encoded.describe())
```

```python
# Check highest price again
top_10_prices = df_encoded["PRICE"].nlargest(20)
print(top_10_prices)
```

```python
# Save the modified DataFrame to a new CSV file
output_file_path = "./Clean_NY.csv"
df_encoded.to_csv(output_file_path, index=False)
```
